```python
# Import PyTorch
import torch

# Create a vector
v = torch.tensor([10, 20, 30], dtype=torch.float32)

# Create a constant vector
const_vector = torch.tensor([5, 5, 5], dtype=torch.float32)

# Perform operations
addition = v + const_vector
subtraction = v - const_vector
multiplication = v * const_vector
division = v / const_vector

# Print results
print("Original vector:", v)
print("Constant vector:", const_vector)

print("\nAddition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```
