```python
# Import libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Load dataset - Changed to use an available sample dataset
data = pd.read_csv('/content/sample_data/california_housing_train.csv')

# Select a numerical column for demonstration
temp_data = data['housing_median_age'].values
temp_data = temp_data.reshape(-1, 1)

# Normalize data
scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(temp_data)

# Create input-output sequences
x = []
y = []

# Ensure there's enough data for the sequence length (30)
sequence_length = 30
if len(data_scaled) < sequence_length + 1:
    print(f"Not enough data for sequence length {sequence_length}. Please use a larger dataset.")
else:
    for i in range(sequence_length, len(data_scaled)):
        x.append(data_scaled[i-sequence_length:i, 0])
        y.append(data_scaled[i, 0])

    x = np.array(x)
    y = np.array(y)

    # Reshape for LSTM [samples, time_steps, features]
    x = x.reshape(x.shape[0], x.shape[1], 1)

    # Build LSTM model
    model = Sequential([
        LSTM(50, activation='relu', input_shape=(x.shape[1], 1)),
        Dense(1)
    ])

    # Compile model
    model.compile(optimizer='adam', loss='mse')

    # Train model
    model.fit(x, y, epochs=20, batch_size=16, verbose=0)

    # Prediction
    pred = model.predict(x)

    print("Prediction:\n", pred[:2])

    # Example output
    # [[29.85],
    #  [30.12]]

    print("Result: Successful implementation of LSTM model with sample data.")
```
